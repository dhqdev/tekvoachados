```python
# Versão em Notebook do Fluxo de Trabalho
Este notebook transforma um fluxo de trabalho encurtado em células passo a passo. Ele segue a estrutura pedida: importação, carregamento/preparação, definições, execução e visualização.
</VSCode.Cell>

<VSCode.Cell language="python">
# Importar bibliotecas necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print('Bibliotecas carregadas com sucesso.')
</VSCode.Cell>

<VSCode.Cell language="markdown">
## 2. Carregar e Preparar Dados
Aqui simulamos a carga de dados que servem como base para a análise ou processamento no notebook.
</VSCode.Cell>

<VSCode.Cell language="python">
# Carregar dados de exemplo
sample_data = {
    'produto': ['Tênis', 'Perfume', 'Monitor', 'Whey', 'Churrasqueira'],
    'preco_original': [599, 349, 1299, 189, 599],
    'preco_desconto': [369, 206, 919, 127, 289],
    'desconto': [38, 41, 29, 33, 52]
}

df = pd.DataFrame(sample_data)

df['economia'] = df['preco_original'] - df['preco_desconto']

df.head()
</VSCode.Cell>

<VSCode.Cell language="markdown">
## 3. Definir Funções e Configurações
Definimos funções reutilizáveis para processar os dados e aplicar regras de desconto ou urgência.
</VSCode.Cell>

<VSCode.Cell language="python">
# Definir funções reutilizáveis

def calcular_desconto(row):
    return f"{row['desconto']}% off"


def marcar_urgencia(df, limite=30):
    df['urgencia'] = np.where(df['desconto'] >= limite, '🔥 Esgotando rápido', '👍 Oferta valida')
    return df

# Aplicar função aos dados
mana_df = df.copy()
mana_df['tag_desconto'] = mana_df.apply(calcular_desconto, axis=1)
mana_df = marcar_urgencia(mana_df)

mana_df.head()
</VSCode.Cell>

<VSCode.Cell language="markdown">
## 4. Executar Passo a Passo no Notebook
Agora executamos o processamento passo a passo e validamos os resultados com saídas claras.
</VSCode.Cell>

<VSCode.Cell language="python">
# Validação e inspeção de resultados
print('Total de itens: ', len(mana_df))
print('Itens com urgência alta:')
print(mana_df[mana_df['urgencia'] == '🔥 Esgotando rápido'])
</VSCode.Cell>

<VSCode.Cell language="markdown">
## 5. Visualizar e Salvar Resultados
Exibimos gráficos simples e salvamos o resultado final em um arquivo CSV para uso posterior.
</VSCode.Cell>

<VSCode.Cell language="python">
# Visualização simples
plt.figure(figsize=(8, 4))
plt.bar(mana_df['produto'], mana_df['desconto'], color=['#8424F0', '#A855F7', '#7824E4', '#B57DFF', '#3C0C78'])
plt.title('Desconto por Produto')
plt.ylabel('Desconto (%)')
plt.xlabel('Produto')
plt.grid(axis='y', alpha=0.3)
plt.show()

# Salvar resultados
mana_df.to_csv('resultados_desconto.csv', index=False)
print('Arquivo resultados_desconto.csv salvo com sucesso.')
```
